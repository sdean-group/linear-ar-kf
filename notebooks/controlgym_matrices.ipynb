# ControlGym Matrices

This notebook prints the actual matrices behind the environments in `controlgym`.

It shows:
- the discrete-time matrices used by `controlgym.make(...)`
- the derived quadratic cost matrices `Q`, `R`, and `S`


```python
from pathlib import Path
import importlib.resources as pkg_resources

import numpy as np
import scipy.io as sio
from controlgym.envs import linear_control_src, make

np.set_printoptions(precision=5, suppress=True, linewidth=160)
```

```python
env_id = "ac6"
sample_time = 0.1
noise_cov = 0.1
random_init_state_cov = 0.1
```

```python
def show_matrix(name, matrix):
    matrix = np.asarray(matrix)
    print(f"{name}: shape={matrix.shape}")
    print(matrix)
    print()
```

```python
with pkg_resources.path(linear_control_src, f"{env_id}.mat") as mat_path:
    mat_path = Path(mat_path)
    raw = sio.loadmat(mat_path)

print(f"Loaded raw file: {mat_path}")
print()

for key in ["A", "B1", "B2", "C", "C1", "D11", "D12", "D21", "x_0"]:
    if key in raw:
        show_matrix(f"raw[{key}]", raw[key])
```

```python
env = make(
    env_id,
    sample_time=sample_time,
    noise_cov=noise_cov,
    random_init_state_cov=random_init_state_cov,
)

print(f"Instantiated controlgym env: {env_id}")
print(f"sample_time={env.sample_time}")
print(f"noise_cov={env.noise_cov}")
print(f"random_init_state_cov={env.random_init_state_cov}")
print(f"n_state={env.n_state}, n_disturbance={env.n_disturbance}, n_action={env.n_action}, n_observation={env.n_observation}")
print(f"spectral_radius(A)={np.max(np.abs(np.linalg.eigvals(env.A))):.6f}")
```

```python
show_matrix("env.A", env.A)
show_matrix("env.B1", env.B1)
show_matrix("env.B2", env.B2)
show_matrix("env.C", env.C)
show_matrix("env.C1", env.C1)
show_matrix("env.D11", env.D11)
show_matrix("env.D12", env.D12)
show_matrix("env.D21", env.D21)
```

```python
show_matrix("env.Q", env.Q)
show_matrix("env.R", env.R)
show_matrix("env.S", env.S)
show_matrix("env.init_state", env.init_state)
```
